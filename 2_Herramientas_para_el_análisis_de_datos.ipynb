### Manejo de matrices

```python
# Reshape()
array = np.arange(12)
print("Vector original:", array)
# Reformar el array a una matriz de 4x3
nuevo_array = array.reshape(4,3)
print("Matriz generada: ","\n", nuevo_array)
```

```python
# Flatten()
print("Forma de la matriz:", nuevo_array.shape)
vector = nuevo_array.flatten()
print("Vector: ",vector)
print("Forma del vector ", vector.shape)
```

```python
# Traspose()
matriz = nuevo_array.transpose()
print("Matriz transpuesta: \n", matriz)
```

```python
# Concatenate() horizontal
A = np.array([[1,2],[3,4]])
print("Matriz A: \n", A)
B = np.array([[5,6],[7,8]])
print("Matriz B: \n", B)
C = np.concatenate((A,B), axis = 1)
print("Matriz C: \n", C)
```

```python
# hstack()
A = np.array([[1,2],[3,4]])
print("Matriz A: \n", A)
B = np.array([[5,6],[7,8]])
print("Matriz B: \n", B)
C = np.hstack((A,B))
print("Matriz C: \n", C)
```

```python
# Concatenate() vertical
A = np.array([[1,2],[3,4]])
print("Matriz A: \n", A)
B = np.array([[5,6],[7,8]])
print("Matriz B: \n", B)
C = np.concatenate((A,B), axis = 0)
print("Matriz C: \n", C)
```

```python
# vstack()
A = np.array([[1,2],[3,4]])
print("Matriz A: \n", A)
B = np.array([[5,6],[7,8]])
print("Matriz B: \n", B)
C = np.vstack((A,B))
print("Matriz C: \n", C)
```

```python
# column_stack()
arr1 = np.arange(1,5)
print("Array 1: \n", arr1)
arr2 = 2 * arr1
print("Array 2: \n", arr2)
n = np.column_stack((arr1,arr2))
print("Matriz apilada: \n", n)
```

```python
# row_stack()
arr1 = np.arange(1,5)
print("Array 1: \n", arr1)
arr2 = 2 * arr1
print("Array 2: \n", arr2)
n = np.row_stack((arr1,arr2))
print("Matriz apilada: \n", n)
```

```python
# np.dstack()
'''El apilamiento en profundidad con NumPy se realiza mediante la función np.dstack(),
la cual combina matrices apilándolas a lo largo del tercer eje (eje 2),
transformando vectores o tablas en una estructura tridimensional.'''

# Crear dos matrices tridimensionales
arr1 = np.array([[[1,2],[3,4],[5,6],[7,8]]])
arr2 = np.array([[[9,10],[11,12],[13,14],[15,16]]])

# Apilar las matrices en profundidad
matriz_final = np.dstack((arr1, arr2))
print("Matriz 1:")
print(arr1)
print("Matriz 2:")
print(arr2)
print("Matriz final:")
print(matriz_final)
```

### Segmentación de matrices

```python
# Corte de la posición 2 a la 5
arr = np.array([1,2,3,4,5,6,7])
print(arr[2:5])

# Corte de la posición 2 a la última
print(arr[2:])

# Corte de la posición -2 a la última
print(arr[-2:])

# Corte de la posición 3 a la 8 con incremento de 2
print(arr[3:8:2])
```

```python
# Indexación booleana
edades = np.arange(1,21)
print("Vector original:\n", edades)
print("Después de la condición", edades[edades%2==0])
```

```python
# Selección de filas específicas
arr = np.arange(1,21).reshape(5,4)
print("Array original:\n", arr)

# Selección de la fila 0 y 1
indices = [0,1]
print("Selección de la fila 0 y 1:\n", arr[indices])

# Selección de la fila 2 y 3
indices = [2,3]
print("Selección de la fila 2 y 3:\n", arr[indices])
```

### Otras operaciones sobre matrices

```python
# Suma, resta, multiplicación, división y producto punto

# Crear dos matrices
A = np.arange(1,5).reshape(2,2)
print("Matriz A: \n", A)
B = np.arange(5,9).reshape(2,2)
print("Matriz B: \n", B)

# Suma
print("Suma: \n", A + B)

# Resta
print("Resta: \n", A - B)

# Multiplicación
print("Multiplicación: \n", A * B)

# División
print("División: \n", A / B)

# Producto punto
print("Producto punto: \n", np.dot(A,B))
```

```python
## Dataframes de ejemplo - estudiantes
datos_periodo1 = {
    'ID':[1,2,3,4,5],
    'Nombre':['Juan','Maria','Pedro','Laura','Carlos'],
    'Calificacion':[85,90,78,92,88]
}

df_periodo1 = pd.DataFrame(datos_periodo1)

datos_periodo2 = {
    'ID':[2,3,4,5,6,7],
    'Nombre':['Maria','Pedro','Laura','Carlos','Sofia','Luis'],
    'Calificacion':[88,22,77,91,89,93]
}

df_periodo2 = pd.DataFrame(datos_periodo2)
```

```python
# Inner join
union_estudiantes = pd.merge(df_periodo1, df_periodo2,on='ID', suffixes=('_periodo1','_periodo2'), how='inner')
union_estudiantes
```

```python
# Left join
union_estudiantes = pd.merge(df_periodo1, df_periodo2,on='ID', suffixes=('_periodo1','_periodo2'), how='left')
union_estudiantes
```

```python
# Right join
union_estudiantes = pd.merge(df_periodo1, df_periodo2,on='ID', suffixes=('_periodo1','_periodo2'), how='right')
union_estudiantes
```

```python
# Outer join
union_estudiantes = pd.merge(df_periodo1, df_periodo2,on='ID', suffixes=('_periodo1','_periodo2'), how='outer')
union_estudiantes
```

```python
df_concatenado1 = pd.concat([df_periodo1, df_periodo2], axis=0)
print(f'Concatenado horizontal: \n{df_concatenado1}')

df_concatenado2 = pd.concat([df_periodo1, df_periodo2], axis=1)
print(f'Concatenado vertical: \n{df_concatenado2}')
```

## Matplotlib

```python
import matplotlib.pyplot as plt
```

### Gráficos de lineas

```python
X = np.arange(1,20)
# Grafico de linea
plt.plot(X, X+1, label='linear')
plt.show()
```

```python
# Grafico de dos lineas

## Primera linea
x1 = [1,3,5,7,9,11]
y1 = [10,25,35,33,41,59]

## Segunda linea
x2 = [2,4,6,8,10,12]
y2 = [15,29,32,33,38,55]

## Etiquetas de eje
plt.xlabel('Eje X')
plt.ylabel('Eje y')

## Título del gráfico
plt.title('Gráfico de dos lineas')

## Leyendas
plt.legend()

## Dibujo del gráfico
plt.plot(x1,y1, label='Serie 1', color='green')
plt.plot(x2,y2, label='Serie 2', color='purple')
plt.show()
```

### Gráficos de dispersión

```python
x = [1,3,5,7,9,11]
y = [10,25,35,33,41,59]

plt.xlabel('Eje X')
plt.ylabel('Eje y')
plt.title('Gráfico de dispersión')
plt.scatter(x,y, c='red', marker='*', alpha=0.5)
plt.show()
```

### Gráfico circular

```python
asignaturas = ['Matemáticas','Programación','Algebra','Cálculo']
notas = [75,90,85,85]

# Dibujo del gráfico
colores = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
plt.pie(notas, labels=asignaturas, colors=colores, startangle=90, shadow=True, explode=(0,0.1,0,0),
        autopct='%1.1f%%', textprops={'color':'white', 'size':9, 'weight':'bold'})
plt.title('Notas de los estudiantes en el examen')
plt.show()
```

### Gráfico de barras

```python
semestres = [1,2,3,4,5]
cantidad_estudiantes = [21,40,22,29,32]

# Dibujo del gráfico
plt.bar(semestres, cantidad_estudiantes, color='red')
plt.xlabel('Semestre')
plt.ylabel('Cantidad de estudiantes')
plt.title('Cantidad de estudiantes por semestre')
plt.show()
```

### Histograma

```python
edades = [21,28,32,34,35,35,37,42,47,55]
grupos = [20,30,40,50,60]

# Dibujo del gráfico
plt.hist(edades, bins=grupos, color='green', rwidth=0.6)
plt.xlabel('Edad')
plt.ylabel('Frecuencia')
plt.title('Distribución de edades por grupo')
plt.show()
```
